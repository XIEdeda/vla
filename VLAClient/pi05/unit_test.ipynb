```python
import numpy as np

np.random.seed(42)  # 保证可复现

test_cases = []

# 1. 线性上升（斜坡）
test_cases.append(np.linspace(0.0, 2.0, 20).tolist())

# 2. 阶跃信号（前10个0，后10个1）
test_cases.append([0.0]*10 + [1.0]*10)

# 3. 正弦波
test_cases.append((np.sin(np.linspace(0, 4*np.pi, 20)) * 0.5 + 0.5).tolist())

# 4. 随机噪声（小波动）
test_cases.append((0.5 + np.random.normal(0, 0.05, 20)).tolist())

# 5. 大突变（中间跳变）
test_cases.append([0.0]*9 + [2.0] + [0.0]*10)

# 6. 平台+斜坡
test_cases.append([0.2]*5 + np.linspace(0.2, 1.5, 15).tolist())

# 7. 递减斜坡
test_cases.append(np.linspace(1.5, 0.0, 20).tolist())

# 8. 高频震荡
test_cases.append((np.sin(np.linspace(0, 12*np.pi, 20)) * 0.3 + 0.5).tolist())

# 9. 全零
test_cases.append([0.0] * 20)

# 10. 指数逼近（模拟真实控制器响应）
t = np.arange(20)
test_cases.append((1 - np.exp(-t / 5.0)).tolist())
```
